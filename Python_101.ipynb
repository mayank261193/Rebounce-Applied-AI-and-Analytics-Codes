**Problem 2-The Currency Exchange**

```python
Value=75000                                                                      #Assign value to variable value
Amount_INR ="₹"+str(Value)                                                       #Assign value to string and concatenate
Ex_Rate=str("1")+str("₹")+" ""="" "+str("1.75")+"JPY"                            #define Exchange rate
Exchange=int(Value*1.75)
Amount_JPY=f"{Exchange:.2f}"+"JPY"                                               #Using f string for decimal format
print("i)  Amount in Indian currency  =",Amount_INR)
print("ii) Ex_rate for Japan>>..  ",Ex_Rate)
print("iii)Amount in Japanese currency=",Amount_JPY)
```

**Problem 5-The Password Strength Check**

**Problem 9-Sort Products by Price**
