**Problem 2-The Currency Exchange**

```python
Value=75000                                                                      #Assign value to variable value
Amount_INR ="₹"+str(Value)                                                       #Assign value to string and concatenate
Ex_Rate=str("1")+str("₹")+" ""="" "+str("1.75")+"JPY"                            #define Exchange rate
Exchange=int(Value*1.75)
Amount_JPY=f"{Exchange:.2f}"+"JPY"                                               #Using f string for decimal format
print("Amount in Indian currency=",Amount_INR)
print("Exchange rate for Japan>>",Ex_Rate)
print("Amount in Japanese currency=",Amount_JPY)
```

**Problem 3-The Grocery Bill**

```python
B=40                                                                                              #Assign value to variable B
M=65.50                                                                                           #Assign value to variable M
E=120                                                                                             #ssign value to variable E
S=B+M+E                                                                                           #Adding all the values assigned
Sum=f"{S:.2f}"
print("Bread_price\t=","₹",str(B),"\nMilk_price\t=","₹",str(M),"\nEggs_price\t=","₹",str(E),"\t")
print("Total price\t=","₹",Sum)
A=S/3                                                                                             #Take average of Sum of values
Average=f"{A:.2f}"                                                                                #Formating to two decimal place
print("Average\t\t=","₹",Average)
```

**Problem 5-The Password Strength Check**

```python
P="Time@123"
if len(P)>=8 and '@' in P:
    print(True,"and Strong password")
elif len(P)>=8 and '@' not in P:
    print(True,"and Weak password")
else:
    print(False)
```
