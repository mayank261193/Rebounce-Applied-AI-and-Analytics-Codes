# **Retail Store data analysis**

### **Import Libraries**

````python
import pandas as pd
import numpy as np
````

### **Load Dataset**

````python
#Using url to read file
url = "https://huggingface.co/datasets/aarav912/online-retail/resolve/main/online_retail.csv"
df = pd.read_csv(url)
````

### **Do columns Need cleaning?**
No columns needs cleaning as empty cells could filled with every data update.


### **Clean the data**

````python
df.dropna(subset=['CustomerID'], inplace=True)

print(f"Missing CustomerID values after drop:{df['CustomerID'].isna().sum()}")
print(f"Shape after dropping:{df.shape}")
# I opted to drop the missing values because without cutomer ID data is useless.
````

````python
#Convert InvoiceDate type into datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df.dtypes
````

````python
#Dropping duplicate rows
df = df.drop_duplicates()
df.shape
````

````python
#Remove rows where Quantity is zero or negative, or where UnitPrice is zero or negative
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]
df.shape
````

### **Create Sales column**

````python
# Create a new column: Sales = Quantity × UnitPrice.
df["Sales"] = df["Quantity"] * df["UnitPrice"]
df.shape
````

### **Numpy Analysis**

````python
#Find statistical parameters usign inbuilt functions
import numpy as np

sales = df["Sales"].to_numpy()

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))
print("Standard Deviation:", np.std(sales))
#The data is right skewed as mean>median which means majority of sales below median.
````

### **Pandas filtering**



````python
#Sales above 500
count = (df["Sales"] > 500).sum()
print("Transactions above 500:", count)
````

````python
#10 largest sales
top_10 = df.sort_values("Sales", ascending=False).head(10)
print(top_10)
````

````python
#transactions come from Germany
germany = df[df["Country"] == "Germany"]
print(germany)
````

````python
#transactions have a Quantity above 1,000
transactions = df[df["Quantity"] > 1000]
print(transactions)
#They are wholesale orders hence data seems believable
````

### **GroupBy and aggregation**

````python
#Data summary
summary = df.groupby("Country").agg(
    Number_of_Transactions=("Country", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean")
).reset_index().sort_values("Total_Sales", ascending=False)

print(f"{summary}\n The country with the biggest total is rarely the one with\n the biggest average because the number of transactions are varying directly \nwith sales hence average country with biggest total sale is minimum")


````

### **Merge/region analysis**

````python
#create dataframe from dictionary
region_lookup = {    "United Kingdom": "Europe",
                 "France": "Europe",    "Germany": "Europe",
                     "Spain": "Europe",    "Netherlands": "Europe",
                     "Belgium": "Europe",    "Switzerland": "Europe",
                     "Portugal": "Europe",    "Australia": "Oceania",
                     "Japan": "Asia",    "USA": "Americas",
                }
region_df = pd.DataFrame(list(
    region_lookup.items()),
    columns=["Country", "Region"]
)

print(region_df)
````

````python
#Left merge
summary = pd.merge(
    summary,
    region_df,
    on="Country",
    how="left"
)

summary["region_df"] = summary["Region"].fillna("Other")

print(summary)
````

````python
#Group by region
region_sales = (
    summary.groupby("Region")["Total_Sales"]
    .sum()
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

print(region_sales)
````

### **Final Finding**



### 1. Important finding
```
**The country with the highest total sales is not necessarily
the country with the highest average sales.**Total sales depend
heavily on the number of transactions, while average sales shows
the typical value of an individual transaction.
```

### 2. One comparison
```
If the mean is noticeably higher than the median, it suggests that
some high-value transactions are pulling the average upward
```

### 3. Observation about the data
```
**Sales are likely right-skewed**
This would explain why the mean is higher than the median and
indicates that the typical transaction is better represented
by the median.
```
### 4. Limitation
```
**The data required cleaning before analysis.** Duplicate rows were
removed, and transactions with `Quantity <= 0` or `UnitPrice <= 0`
were excluded because they represented cancellations or adjustments
```
