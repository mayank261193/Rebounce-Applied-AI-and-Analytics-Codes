```python
#Readme
#Track-A
#Track Name-Personal Expense tracker
#Description-A menu-driven tool that tracks daily expenses and shows totals by category
#How to Run this program-
```

```python
#Personal expense tracker
# Daily Expense Tracker

expenses = [] # Initialize an empty list to store expenses

while True:

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        category = input("Enter Category: ")
        amount = float(input("Enter Amount: "))
        description = input("Enter Description: ") # Corrected variable name from 'date' to 'description'

        expense = {
            "Category": category,
            "Amount": amount,
            "Description": description,
        }

        expenses.append(expense)
        print("Expense added successfully!")

    #.......................................
    elif choice == "2":
        if len(expenses) == 0: # Check the 'expenses' list, not 'expense' dictionary
            print("No expenses recorded.")
        else:
            print("\n--- All Expenses ---")
            for exp in expenses: # Iterate through the 'expenses' list
                print(exp["Category"],exp["Amount"],exp["Description"])

    #............................................
    elif choice == "3":
        if len(expenses) == 0: # Check the 'expenses' list
            print("No expenses recorded.")
        else:
            category_totals = {}

            for exp in expenses: # Iterate through the 'expenses' list
                category = exp["Category"]
                amount = exp["Amount"]

                if category in category_totals:
                    category_totals[category] += amount
                else:
                    category_totals[category] = amount

            print("\n--- Category-wise Total ---")
            for category, total in category_totals.items():
                print(f"{category}: ₹{total:.2f}")

    #.....................................................
    elif choice == "4":
        print("Have a nice day!")
        break

    else:
        print("Invalid choice! Please enter 1-4.")

```
