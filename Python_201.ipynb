1. Add expense
2. View all expenses
3. Calculate total
4. Calculate total by category
5. Save expenses
6. Load expenses
7. Keep menu running until Quit

```python
%%writefile Expense.py
#Define List
expenses = []
#Add Expense Function
def add_expense(expenses):

    category = input("Enter category: ")
    amount = float(input("Enter amount: "))
    description = input("Enter description: ")

    expense = {
        "category": category,
        "amount": amount,
        "description": description
    }

    expenses.append(expense)

    print("Expense added successfully!")
#view all expense
def view_expenses(expenses):

    if not expenses:
        print("No expenses found.")
        return

    print("\n----- ALL EXPENSES -----")

    for i, expense in enumerate(expenses, start=1):

        print(
            f"{i}. "
            f"{expense['category']} | "
            f"Rs {round(expense['amount'],2)} | "
            f"{expense['description']}"
        )
#Total spending
def total_expenses(expenses):

    total = sum(expense["amount"] for expense in expenses)

    print(f"\nTotal spent: Rs {round(total,2)}")
#Total by category
def category_totals(expenses):

    totals = {}

    for expense in expenses:

        category = expense["category"]
        amount = expense["amount"]

        if category in totals:
            totals[category] += amount
        else:
            totals[category] = amount

    print("\n----- SPENDING BY CATEGORY -----")

    if not totals:
        print("No expenses found.")
        return

    for category, amount in totals.items():

        print(f"{category}: Rs {round(amount,2)}")
#Menu
while True:

    print("\n==============================")
    print(">>>>>EXPENSE TRACKER<<<<<<")
    print("==============================")

    print("a). Add expense")
    print("b). View all expenses")
    print("c). View total spending")
    print("d). View spending by category")
    print("e). Quit")

    choice = input("\nEnter your choice: ")


    if choice == "a":

        add_expense(expenses)


    elif choice == "b":

        view_expenses(expenses)


    elif choice == "c":

        total_expenses(expenses)


    elif choice == "d":

        category_totals(expenses)


    elif choice == "e":

        print("Come again!")

        break


    else:

        print("wrong choice. Please try again.")
```
